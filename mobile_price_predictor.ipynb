```python
df['Storage'] = pd.to_numeric(
    df['Storage'].astype(str).str.extract(r'(\d+)', expand=False),
    errors='coerce'
)
df['RAM'] = pd.to_numeric(
    df['RAM'].astype(str).str.extract(r'(\d+)', expand=False),
    errors='coerce'
)
df.head()
```

```python
df['Camera (MP)'] = pd.to_numeric(
    df['Camera (MP)'].astype(str).str.extract(r'^\s*(\d+)', expand=False),
    errors='coerce'
)
df.head()
```

```python
df['Screen Size (inches)'] = pd.to_numeric(df['Screen Size (inches)'],errors='coerce')
```

```python
df.info()
```

```python
df['Price ($)'] = pd.to_numeric(
    df['Price ($)'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False),
    errors='coerce'
)
df.head(50)
```

```python
df = df.drop(columns=['Model'])
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['Brand'] = encoder.fit_transform(df['Brand'])
```

```python
x = df.drop(columns = ['Price ($)'])
y = df['Price ($)']
```

```python
y.info()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model = RandomForestRegressor(n_estimators=100)
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)

print(f"MSE  {mse}")
print(f"MAE  {mae}")
print(f"R2  {r2}")
```

```python
import joblib
import joblib

joblib.dump({
    'model': model,
    'encoder': encoder,
    'feature_columns': list(x_train.columns)
}, 'mobile_price_model.pkl')
```
